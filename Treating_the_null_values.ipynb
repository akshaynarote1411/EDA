- bfill : backfill / bfill: use next valid observation to fill gap means it will fill the null value by exactly next(just after) value(observation) of that null from the same column in which the null from.

#### Filling Null with Bivariate Analysis

- Always We should not replace the Null value with mean, median, mode by considering only one column.

- Some time we have to consider the other columns of dataset also to get more accurate value to replace Null value.

- below is the example on that

```python
d = {"Education":["mbbs","engg","mbbs","engg","engg"],"Salary":[70,40,80,50,np.nan]}

df5 = pd.DataFrame(d)
df5
```

```python
df5["Salary"].fillna(df5["Salary"].mean())   # It will not gives more approprite value to replace missing value beacuse it will consider all the values in "Salary" column while calculating mean
                                             # But there are different categeries in "Education" column which affect the "Salary" column's values completely
                                             # So have to consider that categories and by making group of similar categories we should fill the missing values from "Salary" colm.
                                             # Ex. In "Education" colm 'mbbs' & 'engg', two categories are there which are completely different and and which will impact on their salary, so whilw calculating engg's salary we shold noy conider the mbbs category.
```

```python
# So the correct way is 

mean = df5.groupby(["Education"])["Salary"].mean()
print(mean)

df5["Salary"].fillna(mean[0])
```

#### linear Interpolation to fill the Null :

```python
df6 = pd.DataFrame([10,np.nan,np.nan,40,50], columns=["Values"])
df6
```

```python
df6["Values"].interpolate(method="linear")   
```

#### interpolate(method="linear")

- It is used to replace the null values by considering the side by values(previous & next values of null value)


- It will calculate the range between tha last(previous) & next values of the null value(s) and count the number of missing value(s) in that range.

- Then it will divide that range value by the no. of missing values plus 1, And then replace the missing values by adding that number to the previos value of that perticluar missing value.


- Ex.
     - In above dataframe,
       
       - range = 40-10 (40 is next value of missing value and 10 is previous. )
             = 30
   
       - no. of missing values = 2 
       
       - adding value = 30/(2+1) = 10
       
       - so 1st null value is replaced by : previous value + adding value = 10 + 10 = 20
       
       - 2nd null value is replaced by : it's previuos value(i.e. 1st null) + adding_value = 20 + 10 = 30
       
       - beacuse of this we get missing values 20 & 30.

```python
# But the linear interpolation is not properly work when we have misiing value at the last of columns.
# It will replace the that last missing value with the previous valid column value.

df6 = pd.DataFrame([10,np.nan,np.nan,40,np.nan,50,np.nan], columns=["Values"])
df6
```

```python
df6["Values"].interpolate(method="linear")    # It will replace the last null with the valid value i.e. 50  
```
