```python
df = pd.read_csv("Visadataset.csv")

df["education_of_employee"].unique()
```

```python
# If we apply Nominal Encoding i.e. LableEncoder on "education_of_employee", it will get fail 
# because  it is "Ordinal Data" means order(priority) between the column category values is very important.
# but nominal encoding can't understand that it encode the cateories by alphabetical order, and gives wrong priority to categories.


from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df["education_of_employee"] = le.fit_transform(df["education_of_employee"])
df
```
