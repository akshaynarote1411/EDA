```python
cat = df.select_dtypes(include="object").columns
cat = cat.drop(cat[0])

d = pd.DataFrame()

for i in cat:
    df2 = pd.DataFrame(zip(df[i].unique(),df.groupby(i).size()),columns=["Labels","Counts"])
    d = pd.concat([d,df2])
d
```

```python
cat = df.select_dtypes(include="object").columns
cat = cat.drop(cat[0])

d = pd.DataFrame()

for i in cat:
    df2 = pd.DataFrame(zip(df[i].unique(),df.groupby(i).size()),columns=["Labels","Counts"])
    d = pd.concat([d,df2],ignore_index=True)
d
```
